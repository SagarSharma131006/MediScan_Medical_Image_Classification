```python
# ================================================================
# MediScan — Day 13
# Cell 5: Log Training & Validation History to W&B
# ================================================================

import pandas as pd
from pathlib import Path

PROJECT = Path("/content/MediScan")
HISTORY_PATH = PROJECT / "reports/day8_final_training_history.csv"

print("=" * 65)
print("MediScan — Logging Training History to W&B")
print("=" * 65)

# ------------------------------------------------
# Load Day 8 training history
# ------------------------------------------------
history_df = pd.read_csv(HISTORY_PATH)

print("\nTraining history loaded successfully.")
print("Shape   :", history_df.shape)
print("Columns :", list(history_df.columns))

print("\nFirst 5 rows:")
display(history_df.head())

# ------------------------------------------------
# Normalize column names for safe matching
# ------------------------------------------------
history_df.columns = [
    col.strip().lower().replace(" ", "_")
    for col in history_df.columns
]

print("\nNormalized columns:")
print(list(history_df.columns))

# ------------------------------------------------
# Identify columns
# ------------------------------------------------
epoch_col = next(
    col for col in history_df.columns
    if col in ["epoch", "epochs"]
)

train_loss_col = next(
    col for col in history_df.columns
    if "train" in col and "loss" in col
)

train_acc_col = next(
    col for col in history_df.columns
    if "train" in col and "acc" in col
)

val_loss_col = next(
    col for col in history_df.columns
    if ("val" in col or "valid" in col) and "loss" in col
)

val_acc_col = next(
    col for col in history_df.columns
    if ("val" in col or "valid" in col) and "acc" in col
)

print("\nDetected columns:")
print("Epoch      :", epoch_col)
print("Train Loss :", train_loss_col)
print("Train Acc  :", train_acc_col)
print("Val Loss   :", val_loss_col)
print("Val Acc    :", val_acc_col)

# ------------------------------------------------
# Log every epoch to W&B
# ------------------------------------------------
for _, row in history_df.iterrows():

    epoch = int(row[epoch_col])

    wandb.log({
        "epoch": epoch,
        "train/loss": float(row[train_loss_col]),
        "train/accuracy": float(row[train_acc_col]),
        "validation/loss": float(row[val_loss_col]),
        "validation/accuracy": float(row[val_acc_col]),
    })

print("\n✓ Training loss logged")
print("✓ Training accuracy logged")
print("✓ Validation loss logged")
print("✓ Validation accuracy logged")
print(f"✓ Total epochs logged: {len(history_df)}")

print("\n" + "=" * 65)
print("✓ Day 8 training history successfully logged to W&B")
print("=" * 65)
```

```python
# ================================================================
# MediScan — Day 13
# Cell 7: Log Per-Class Metrics + ROC-AUC to W&B
# ================================================================

import pandas as pd
from pathlib import Path

PROJECT = Path("/content/MediScan")
REPORTS = PROJECT / "reports"

CLASS_REPORT_PATH = REPORTS / "day9_classification_report.csv"
ROC_PATH = REPORTS / "day9_roc_auc.csv"

print("=" * 65)
print("MediScan — Logging Per-Class Metrics")
print("=" * 65)

# ------------------------------------------------
# Load classification report
# ------------------------------------------------
class_df = pd.read_csv(CLASS_REPORT_PATH)

print("\nClassification report:")
display(class_df)

# ------------------------------------------------
# Load ROC-AUC report
# ------------------------------------------------
roc_df = pd.read_csv(ROC_PATH)

print("\nROC-AUC report:")
display(roc_df)

# ------------------------------------------------
# Log per-class metrics
# ------------------------------------------------

# Known class names
class_names = [
    "glioma",
    "meningioma",
    "notumor",
    "pituitary"
]

for class_name in class_names:

    # Find classification row
    class_row = class_df[
        class_df.iloc[:, 0].astype(str).str.lower() == class_name
    ]

    # Find ROC row
    roc_row = roc_df[
        roc_df.iloc[:, 0].astype(str).str.lower() == class_name
    ]

    if len(class_row) == 0:
        print(f"⚠ Classification metrics not found for {class_name}")
        continue

    precision = float(class_row.iloc[0]["precision"])
    recall = float(class_row.iloc[0]["recall"])
    f1 = float(class_row.iloc[0]["f1-score"])

    log_data = {
        f"class/{class_name}/precision": precision,
        f"class/{class_name}/recall": recall,
        f"class/{class_name}/f1": f1,
    }

    if len(roc_row) > 0:
        # Automatically use the second column as ROC-AUC value
        roc_auc = float(roc_row.iloc[0, 1])
        log_data[f"class/{class_name}/roc_auc"] = roc_auc

    wandb.log(log_data)

    print(
        f"✓ {class_name}: "
        f"P={precision:.4f}, "
        f"R={recall:.4f}, "
        f"F1={f1:.4f}"
        + (
            f", ROC-AUC={roc_auc:.4f}"
            if len(roc_row) > 0
            else ""
        )
    )

print("\n" + "=" * 65)
print("✓ Per-class metrics successfully logged to W&B")
print("=" * 65)
```

```python
# ================================================================
# MediScan — Day 13
# Cell 8: Log Model + Reports as W&B Artifacts
# ================================================================

from pathlib import Path
import wandb

PROJECT = Path("/content/MediScan")
REPORTS = PROJECT / "reports"
FIGURES = PROJECT / "figures"
MODEL_PATH = PROJECT / "models/efficientnet_b0_best_day8.pth"

print("=" * 65)
print("MediScan — W&B Artifact Logging")
print("=" * 65)

# ------------------------------------------------
# Check important artifacts
# ------------------------------------------------
artifact_files = [
    MODEL_PATH,
    REPORTS / "day8_final_configuration.json",
    REPORTS / "day8_final_training_history.csv",
    REPORTS / "day9_evaluation_metrics.json",
    REPORTS / "day9_classification_report.csv",
    REPORTS / "day9_roc_auc.csv",
    FIGURES / "day8_final_loss_curve.png",
    FIGURES / "day8_final_accuracy_curve.png",
    FIGURES / "day9_confusion_matrix.png",
    FIGURES / "day9_roc_curves.png",
]

print("\nArtifact verification:")

existing_files = []

for file_path in artifact_files:
    if file_path.exists():
        print(f"✓ {file_path.name}")
        existing_files.append(file_path)
    else:
        print(f"⚠ Missing: {file_path.name}")

# ------------------------------------------------
# Create W&B artifact
# ------------------------------------------------
artifact = wandb.Artifact(
    name="mediscan-efficientnet-b0-results",
    type="model",
    description="MediScan EfficientNet-B0 final model, evaluation reports and figures."
)

# Add model
if MODEL_PATH.exists():
    artifact.add_file(str(MODEL_PATH), name="efficientnet_b0_best_day8.pth")

# Add reports
for file_path in existing_files:
    if file_path.suffix.lower() in [".csv", ".json", ".md"]:
        artifact.add_file(
            str(file_path),
            name=f"reports/{file_path.name}"
        )

# Add figures
for file_path in existing_files:
    if file_path.suffix.lower() in [".png", ".jpg", ".jpeg"]:
        artifact.add_file(
            str(file_path),
            name=f"figures/{file_path.name}"
        )

# ------------------------------------------------
# Log artifact
# ------------------------------------------------
run.log_artifact(artifact)

print("\n✓ W&B artifact created")
print("Artifact name: mediscan-efficientnet-b0-results")
print("Artifact type: model")

print("\nFiles tracked:")
for file_path in existing_files:
    print("→", file_path.name)

print("\n" + "=" * 65)
print("✓ Model, reports and figures successfully logged as W&B artifact")
print("=" * 65)
```
